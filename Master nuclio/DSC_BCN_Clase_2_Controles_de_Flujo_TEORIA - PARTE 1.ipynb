### Estructuras de repetición (bucles)



- bucle *while* = **mientras** se cumpla cierta condición hazme esto
- bucle *for* = **por cada** elemento en mi objeto, hazme esto


#### Bucle **while**

Utilizado para ejecutar un bloque de código repetidamente mientras una condición específica sea evaluada como verdadera.

"Repita **mientras** condición"

```python
contador = 0
a=0
while (contador < 5) and (a < 4):
    print("El contador es", a)
    #contador += 1  # Actualización de la condición para evitar un bucle infinito
    a +=1

print("Bucle finalizado")
```

```python
#Condición multiple: Acumular una variable y enseñar la cantidad de iteraciones
numero=1
suma=0
vuelta = 0

while (numero <= 10) and (suma < 100):
  vuelta +=1
  suma +=numero
  numero +=1

  print(f"Vuelta: {vuelta}")
  print(f"Suma: {suma}")
  print(f"numero: {numero}")

print("La suma de los números hasta alcanzar 100 es ", suma)
```

#### Bucle **for**

Utilizado para iterar sobre **secuencias** de elementos, como listas, tuplas, diccionarios, cadenas de caracteres, rangos, entre otros.

"Repita **para cada elemento** en mi secuencia"

```python
#Iteración en listas, tuplas y conjuntos
mi_lista = [1, 2, 3, 4, 5]

for x in mi_lista:
  print(x)

```

```python
#Iteración en diccionarios
edades = {"Juan": 30, "María": 25, "Pedro": 35}


```

```python

for x,y in edades.items():
  print(x)
  print(y)

```

```python
#Iterar sobre una cadena de caracteres:
mensaje = "Hola mundo"

for x in mensaje:
  print(x)
```

### Sentencias "break", "continue", "pass"

**break**: salida inmediata de un bucle. El control del programa pasa a la primera instrucción fuera del bucle.

**continue**: salta a la siguiente iteración del bucle sin ejecutar el resto del código dentro del bucle para la iteración actual. Útil para omitir ciertas iteraciones basadas en alguna condición.

**pass**:  No hace nada. Se utiliza cuando la sintaxis requiere una declaración, pero no deseamos ejecutar ninguna operación. Generalmente utilizado como marcador de posición.

```python
#Diferencias prácticas - break
for i in range(1, 6):
    if i % 2 == 0:
        break
    print(i)


```

```python
#Diferencias prácticas - continue
for i in range(1, 6):
    if i % 2 == 0:
        continue
    print(i)
```

```python
#Diferencias prácticas - pass
for i in range(1, 6):
    if i % 2 == 0:
        pass
    print(i)
```

####Retorno de valores

```python
def suma(a, b):
    return a + b

resultado_suma = suma(3, 5)
print("El resultado de la suma es:", resultado_suma)
```

```python
#retorno de más de un valor
def suma_resta(a, b):
    return a + b, a-b

resultado_suma, resultado_resta = suma_resta(3, 5)
print("El resultado de la suma es:", resultado_suma)
print("El resultado de la resta es:", resultado_resta)
```

#### **match()**

Introducida a partir de la versión **Python 3.10**. Proporciona una forma más expresiva y poderosa de hacer coincidir patrones en datos.

Proporciona un poderoso mecanismo para expresar lógica condicional basada en patrones. Útil para una variedad de aplicaciones, como análisis de datos, procesamiento de lenguaje natural y programación funcional.


```python
num1 = int(input("Introduzca el número 1 (entero):"))
num2 = int(input("Introduzca el número 2 (entero):"))
operacion = input("Introduzca la operacion (entero):")

match operacion:
    case '+':
        resultado = num1 + num2
        print(f"{num1} + {num2} = {resultado}")
    case '-':
        resultado = num1 - num2
        print(f"{num1} - {num2} = {resultado}")
    case '*':
        resultado = num1 * num2
        print(f"{num1} * {num2} = {resultado}")
    case '/':
        if num2 != 0:
            resultado = num1 / num2
            print(f"{num1} / {num2} = {resultado}")
        else:
            print("No se puede dividir por cero")
    case _:
        print("Operación no válida")


```
